## Регрессия с категориальными предикторами

Рассмотрим регрессию __log(price) на carat, carat^2, color__

```python
# спецификация модели
mod = ols(formula='np.log(price)~carat+I(carat**2)+color', data=df)
# подгонка модели
res = mod.fit()
```

```python
# параметры подогнанной модели регрессии
res.params
```

Как интерпретировать коэффициенты?

Базовая категория переменной `color` — D (лучший цвет): для неё бинарной переменной в модели нет. `color[T.E]` — бинарная переменная, равная 1 для бриллиантов цвета E и 0 для остальных; так же устроены `color[T.F]`, …, `color[T.J]`.

Коэффициент при `color[T.X]` — разница в `log(price)` между бриллиантом цвета X и бриллиантом базового цвета D при том же `carat`. В процентах: приближённо $100\cdot\beta$, точно $100\cdot(e^{\beta}-1)$.

Пример: коэффициент при `color[T.J]` равен $-0.440$. При прочих равных (при том же `carat`) бриллиант цвета J в среднем дешевле бриллианта цвета D на $100\cdot(1-e^{-0.440})\approx 35.61\%$ (приближённо — на $44.01\%$). Для цвета E ($-0.032$) приближённая и точная оценки почти совпадают: $3.19\%$ и $3.14\%$.

Эффект `carat` нельзя прочитать по одному коэффициенту: в модель входит и `carat`, и `carat**2`, поэтому изменение `log(price)` при увеличении `carat` зависит от самого значения `carat`.
