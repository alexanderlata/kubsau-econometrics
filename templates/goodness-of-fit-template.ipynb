# Степень подгонки модели регрессии

```python
import pandas as pd
import numpy as np

from statsmodels.formula.api import ols

# Не показывать Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

```python
# импорт данных
df = pd.read_csv('../datasets/Filename.csv')
```

## Спецификация и подгонка

Рассмотрим регрессию

$$\text{наберите уравнение регрессии}$$

```python
# спецификация модели
mod = ols(formula='your specification', data=df)
# подгонка модели
res = mod.fit()
```

```python
# параметры подогнанной модели регрессии
res.params
```

## Суммы квадратов

```python
# TSS
res.centered_tss
```

```python
# ESS
res.ess
```

```python
# RSS
res.ssr
```

## Коэффициенты $R^2$ и $adj.R^2$

```python
# R^2
res.rsquared
```

```python
# adj R^2
res.rsquared_adj
```

```python
# стандартная ошибка регрессии
res.scale**0.5
```

```python
# стандартная ошибка регрессии (другой способ)
res.mse_resid**0.5
```

## Как отфильтровать определенные строки из DataFrame?

Выбор строк набора данных по условию

Чтобы выбрать строки на основе условия, поместите условие в скобки выбора `[]`

```python
df_new = df[df['exog1']==1]
# если не хотим наследовать индекс, то используем код ниже
# df_new = df[df['exog1']==1].reset_index(drop=True)
```

Условие `df['exog1']==1` внутри скобок выбора проверяет, для каких строк столбец `exog1` имеет значение равное 1

```python
# сброс и удаление старого индекса в DataFrame
# drop=True запрещает pandas сохранять исходный индекс в виде столбца в DataFrame
# inplace=True позволяет pandas заменить индекс в исходном DataFrame вместо создания копии DataFrame
df_new.reset_index(drop=True, inplace=True)
```
