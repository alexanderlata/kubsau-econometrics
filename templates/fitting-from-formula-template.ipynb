# Линейная регрессия: подгонка (через формулу)

```python
import pandas as pd
import numpy as np

from statsmodels.formula.api import ols

# Не показывать Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

```python
# импорт данных
df = pd.read_csv('Filename.csv')
```

## Спецификация и подгонка

```python
# спецификация модели через формулу
mod = ols(formula='your specification', data=df)
# подгонка модели
res = mod.fit()
```

```python
# параметры подогнанной модели регрессии
res.params
```

```python
# параметры подогнанной модели регрессии с округлением
res.params.round(3)
```
