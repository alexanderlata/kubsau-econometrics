## Спецификация и подгонка

```python
# спецификация модели через формулу
mod = ols(formula='your specification', data=df)
# подгонка модели с ковариационной матрицей по умолчанию (неробастной)
res = mod.fit()
```

## BP-тест

Тестируем зависимость дисперсии ошибки от регрессоров модели. В качестве аргумента `exog_het` возьмём матрицу регрессионного дизайна $X$

```python
# BP-тест: LM-статистика и её P-значение
lm, lm_pvalue, fvalue, f_pvalue = het_breuschpagan(resid=res.resid, exog_het=mod.exog)
lm, lm_pvalue
```

## White-тест

Тестируем зависимость дисперсии ошибки от регрессоров модели

```python
# White-тест: LM-статистика и её P-значение
lm, lm_pvalue, fvalue, f_pvalue = het_white(resid=res.resid, exog=mod.exog)
lm, lm_pvalue
```
