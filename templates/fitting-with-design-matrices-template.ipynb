# Линейная регрессия: подгонка (через матрицы дизайна)

```python
import pandas as pd
import numpy as np

from statsmodels.api import OLS, add_constant

# Не показывать Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

```python
# импорт данных
df = pd.read_csv('../datasets/Filename.csv')
```

## Матрицы дизайна

```python
# зависимая переменная
y = df['endog']
# матрица регрессоров с константой
X = add_constant(df[['exog1', 'exog2']])
```

## Подгонка модели

```python
# спецификация модели через матрицы дизайна
mod = OLS(endog=y, exog=X, missing='drop')
# подгонка модели
res = mod.fit()
```
