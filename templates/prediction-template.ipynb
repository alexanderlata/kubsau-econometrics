# Прогнозирование (out-of-sample)

```python
import pandas as pd
import numpy as np

from statsmodels.formula.api import ols

# Не показывать Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

```python
# импорт данных
df = pd.read_csv('../datasets/Filename.csv')
```

## Спецификация и подгонка

```python
# спецификация модели через формулу
mod = ols(formula='your specification', data=df)
# подгонка модели
res = mod.fit()
```

```python
# параметры подогнанной модели регрессии с округлением
res.params.round(2)
```

## Прогнозирование

```python
# создадим датафрейм с новыми данными регрессоров для прогноза
new_data = pd.DataFrame({'exog1': [ ], 'exog2': [ ]})
new_data
```

```python
# прогноз
res.predict(exog=new_data, transform=True).round(2)
# если зависимая переменная в логарифмах:
# np.exp(res.predict(exog=new_data, transform=True)).round(2)
```
